# Dataset inventory

This notebook creates a reproducible, technical inventory. It does not infer plant traits or subjective scene labels.

```python
from pathlib import Path
import sys
ROOT = Path.cwd().resolve()
if not (ROOT / 'src').exists(): ROOT = ROOT.parent
sys.path.insert(0, str(ROOT))
RAW = ROOT / 'data/raw/first_sample'
RESULTS = ROOT / 'results'
```

## Build the inventory

Filename timestamps are parsed when available. EXIF capture time is reported separately. The manual-assessment columns are purposely blank.

```python
from src.dataset_utils import build_inventory, duplicates_by_hash
inventory = build_inventory(RAW)
inventory.to_csv(RESULTS / 'dataset_inventory.csv', index=False)
inventory.head()
```

```python
duplicates = duplicates_by_hash(inventory)
print(f'Images: {len(inventory)}')
print(f'Unreadable: {(~inventory.Is_Readable).sum()}')
print(f'Byte-identical duplicate records: {len(duplicates)}')
print(inventory[['Width_px', 'Height_px', 'File_Extension']].drop_duplicates().to_string(index=False))
```
